## **Câu hỏi 1**

**Giải thuật Support Vector Machine hoạt động như thế nào? Hãy giải thích khái niệm về ranh giới phân
tách (hyperplane) và lề (margin)**

### **Trả lời**

**Support Vector Machine (SVM)** là một thuật toán học máy mạnh mẽ dùng cho phân loại và hồi quy, hoạt động bằng cách tìm một siêu phẳng (hyperplane) tối ưu để phân tách dữ liệu thành các lớp, đồng thời tối đa hóa khoảng cách giữa siêu phẳng và các điểm dữ liệu gần nhất — gọi là lề (margin)

**Cách hoạt động của SVM**
SVM biểu diễn dữ liệu dưới dạng các điểm trong không gian n chiều (n là số đặc trưng). Sau đó, nó tìm một siêu phẳng phân tách (hyperplane) sao cho:

  - Dữ liệu thuộc các lớp khác nhau nằm ở hai phía của siêu phẳng.

  - Khoảng cách từ siêu phẳng đến các điểm gần nhất của mỗi lớp là lớn nhất.

**Giải thích các khái niệm chính**
- **Ranh giới phân tách (Hyperplane)**
  - Là mặt phẳng dùng để phân chia dữ liệu giữa các lớp.

  - Trong không gian 2D → là đường thẳng.

  - Trong không gian 3D → là mặt phẳng.

  - Trong không gian n chiều → là siêu phẳng có (n – 1) chiều.

  - SVM chọn siêu phẳng sao cho nó phân chia dữ liệu tối ưu nhất, tức là có lề lớn nhất.

- **Lề (Margin)**
  - Là khoảng cách từ siêu phẳng đến các điểm dữ liệu gần nhất của mỗi lớp.

  - SVM cố gắng tối đa hóa lề để tăng khả năng tổng quát của mô hình.

  - Các điểm nằm sát lề được gọi là support vectors — chúng là những điểm quan trọng nhất, quyết định vị trí của siêu phẳng.

**Nếu dữ liệu không phân tách tuyến tính**
SVM sử dụng kernel trick để ánh xạ dữ liệu sang không gian đặc trưng cao hơn, nơi có thể tìm được siêu phẳng phân tách tuyến tính. Một số kernel phổ biến:

  - **Linear kernel:** dùng cho dữ liệu tuyến tính.

  - **Polynomial kernel:** ánh xạ dữ liệu theo đa thức.

  - **RBF (Radial Basis Function):** dùng cho dữ liệu phức tạp, không tuyến tính.

**Tóm tắt quy trình SVM**
1. Biểu diễn dữ liệu trong không gian đặc trưng.

2. Tìm siêu phẳng phân tách các lớp.

3. Tối đa hóa lề giữa siêu phẳng và các support vectors.

4. Nếu cần, dùng kernel để ánh xạ dữ liệu sang không gian cao hơn.

**Ưu điểm của SVM:**
- Hiệu quả với dữ liệu có độ phân tách rõ ràng.

- Có thể xử lý dữ liệu phi tuyến bằng kernel.

- Ít bị ảnh hưởng bởi nhiễu nếu chọn đúng kernel và tham số.

**Hạn chế:**
- Tốn thời gian với dữ liệu lớn.

- Khó điều chỉnh nếu dữ liệu không phân tách rõ ràng.

## **Câu hỏi 2**

**Các vector hỗ trợ (support vectors) có vai trò gì trong SVM? Tại sao chúng quan trọng?**

### **Trả lời**

**Support vectors** là những điểm dữ liệu nằm gần nhất với ranh giới phân tách (hyperplane) trong mô hình Support Vector Machine (SVM). Chúng đóng vai trò cốt lõi trong việc xác định vị trí và hướng của siêu phẳng phân tách.

**Vai trò của support vectors trong SVM**
  - Xác định ranh giới phân tách: Chỉ các điểm nằm sát lề (margin) mới ảnh hưởng đến vị trí của siêu phẳng. Những điểm này chính là support vectors.

  - Tối ưu hóa mô hình: SVM tìm siêu phẳng sao cho khoảng cách đến các support vectors là lớn nhất → giúp mô hình tổng quát tốt hơn.

  - Giảm độ phức tạp: Mô hình SVM chỉ cần dựa vào một số ít support vectors, không cần toàn bộ dữ liệu → tiết kiệm tài nguyên tính toán.

  - Tăng độ ổn định: Nếu dữ liệu thay đổi nhưng không ảnh hưởng đến các support vectors, mô hình vẫn giữ nguyên → giúp SVM ổn định hơn.

**Tại sao support vectors quan trọng?**
  - Chúng là những điểm quyết định toàn bộ mô hình SVM.

  - Nếu bạn loại bỏ các điểm không phải support vectors, mô hình vẫn giữ nguyên.

  - Nếu bạn thay đổi một support vector, mô hình sẽ thay đổi theo → cho thấy mức độ ảnh hưởng lớn.

## **Câu hỏi 3**

**Sự khác biệt giữa SVM với lề cứng (hard margin) và lề mềm (soft margin) là gì? Khi nào nên sử dụng lề
mềm?**

### **Trả lời**

**SVM với lề cứng (hard margin)** và lề mềm (soft margin) khác nhau ở cách xử lý dữ liệu không phân tách hoàn toàn:

- **Lề cứng (Hard Margin SVM)**
  - Yêu cầu dữ liệu phân tách hoàn toàn: không có điểm nào nằm sai phía của siêu phẳng.

  - Không cho phép sai số hoặc vi phạm lề.

  - Phù hợp với dữ liệu sạch, không nhiễu, phân lớp rõ ràng.

- **Lề mềm (Soft Margin SVM)**
  - Cho phép một số điểm vi phạm lề hoặc nằm sai phía của siêu phẳng.

  - Dùng tham số C để cân bằng giữa độ rộng của lề và số lượng lỗi.

  - Phù hợp với dữ liệu nhiễu, chồng lấn, không phân tách hoàn toàn.

- **Khi nào nên dùng lề mềm?**
  - Khi dữ liệu thực tế có nhiễu, không thể phân tách hoàn toàn.

  - Khi muốn tăng khả năng tổng quát và tránh overfitting.

  - Khi cần mô hình linh hoạt hơn trong việc xử lý sai số.

**Tóm lại:**

  - Hard margin: chính xác, nhưng chỉ dùng khi dữ liệu phân lớp rõ.

  - Soft margin: phổ biến hơn, xử lý tốt dữ liệu thực tế có nhiễu hoặc phân lớp không hoàn hảo.

## **Câu hỏi 4**

**Hàm nhân (kernel) trong SVM là gì? Hãy giải thích các loại kernel phổ biến (linear, polynomial, RBF) và
khi nào nên sử dụng chúng**

### **Trả lời**

**Hàm nhân (kernel)** trong SVM là một hàm cho phép mô hình phân loại dữ liệu phi tuyến bằng cách ánh xạ dữ liệu từ không gian gốc sang một không gian đặc trưng cao hơn, nơi dữ liệu có thể phân tách tuyến tính. Kernel giúp thực hiện điều này mà không cần tính toán trực tiếp tọa độ trong không gian mới, nhờ vào kỹ thuật gọi là “kernel trick”.

**Các loại kernel phổ biến:**
1. Linear Kernel
   - Công thức: \( K(x, x') = x \cdot x' \)

   - Đặc điểm: Không ánh xạ sang không gian khác, dùng trực tiếp dữ liệu gốc.

   - Khi sử dụng: Khi dữ liệu có thể phân tách tuyến tính hoặc gần tuyến tính. Phù hợp với dữ liệu có số chiều lớn như văn bản, dữ liệu sparse.

2. Polynomial Kernel
   - Công thức: \( K(x, x') = (x \cdot x' + c)^d \)

   - Đặc điểm: Ánh xạ dữ liệu sang không gian đa thức bậc cao.

   - Khi sử dụng: Khi dữ liệu có mối quan hệ phi tuyến dạng đa thức. Có thể điều chỉnh độ phức tạp bằng cách thay đổi bậc 𝑑.

3. RBF (Radial Basis Function) Kernel
   - Công thức: \( K(x, x') = \exp(-\gamma \|x - x'\|^2) \)

   - Đặc điểm: Ánh xạ dữ liệu sang không gian vô hạn chiều, rất linh hoạt.

   - Khi sử dụng: Khi dữ liệu có biên phân tách phức tạp, không rõ dạng. Phù hợp với hầu hết các bài toán phân loại phi tuyến.

**Tổng kết lựa chọn kernel:**
  - **Linear:** Dữ liệu tuyến tính hoặc gần tuyến tính.

  - **Polynomial:** Dữ liệu có quan hệ phi tuyến dạng đa thức.

  - **RBF:** Dữ liệu phi tuyến, biên phân tách phức tạp, không rõ cấu trúc.

## **Câu hỏi 5**

**Tham số C trong SVM có ý nghĩa gì? Nó ảnh hưởng như thế nào đến hiệu suất của mô hình?**

### **Trả lời**

**Tham số C** trong SVM là một siêu tham số điều chỉnh mức độ ưu tiên giữa việc tối đa hóa biên phân tách (margin) và giảm lỗi phân loại trên tập huấn luyện.

1. **Ý nghĩa của C:**
  - C nhỏ → cho phép mô hình chấp nhận một số lỗi phân loại để có biên rộng hơn → giúp tổng quát hóa tốt hơn (giảm overfitting).

  - C lớn → mô hình cố gắng phân loại chính xác tất cả điểm huấn luyện, kể cả điểm nhiễu → dễ overfit nếu dữ liệu có nhiễu.

2. **Ảnh hưởng đến hiệu suất mô hình:**
  - C thấp → mô hình đơn giản hơn, ít nhạy cảm với nhiễu, nhưng có thể underfit nếu biên quá rộng.

  - C cao → mô hình phức tạp hơn, phân loại chính xác hơn trên tập huấn luyện, nhưng dễ overfit và kém tổng quát trên dữ liệu mới.

3. **Tóm lại:**
  - C là công cụ điều chỉnh độ mềm của biên phân tách.

  - Cần tối ưu C bằng kỹ thuật như cross-validation để đạt hiệu suất tốt nhất trên dữ liệu thực tế.

## **Câu hỏi 6**

**Viết đoạn code mẫu bằng Python (sử dụng Scikit-learn) để xây dựng một mô hình SVM cho bài toán
phân loại không? Hãy mô tả các bước thực hiện**

### **Trả lời**

```python
# 1. Import thư viện
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

# 2. Tải dữ liệu Iris
iris = datasets.load_iris()
X = iris.data
y = iris.target

# 3. Chia dữ liệu thành tập huấn luyện và kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 4. Chuẩn hóa dữ liệu
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 5. Khởi tạo và huấn luyện mô hình SVM
model = SVC(kernel='rbf', C=1.0, gamma='scale')  # Có thể thử kernel='linear' hoặc 'poly'
model.fit(X_train, y_train)

# 6. Dự đoán và đánh giá
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
