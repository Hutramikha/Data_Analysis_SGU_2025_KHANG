## **Câu hỏi 1**

**Quy trình khai phá dữ liệu CRISP – DM (Cross Industry Standard Process for Data Mining) là gì ? Quy
trình khai phá dữ liệu SEMMA (Sample, Explore, Modify, Model, Access) là gì?**

### **Trả lời**

CRISP-DM và SEMMA là hai quy trình khai phá dữ liệu phổ biến; cây quyết định là mô hình phân loại trực quan gồm nút gốc, nhánh và nút lá; các tiêu chí phân tách như Gini, Entropy và Information Gain giúp chọn thuộc tính tốt nhất để chia dữ liệu.

CRISP-DM (Cross Industry Standard Process for Data Mining) Là quy trình gồm 6 giai đoạn:

  - Hiểu biết về nghiệp vụ (Business Understanding): Xác định mục tiêu kinh doanh.

  - Hiểu biết về dữ liệu (Data Understanding): Thu thập, khám phá và đánh giá dữ liệu.

  - Chuẩn bị dữ liệu (Data Preparation): Làm sạch, chọn và định dạng dữ liệu.

  - Mô hình hóa (Modeling): Áp dụng thuật toán khai phá dữ liệu.

  - Đánh giá (Evaluation): Kiểm tra mô hình có đạt mục tiêu nghiệp vụ không.

  - Triển khai (Deployment): Đưa mô hình vào sử dụng thực tế.

SEMMA (Sample, Explore, Modify, Model, Assess) được phát triển bởi SAS, gồm 5 bước:

  - Sample: Lấy mẫu dữ liệu đại diện.

  - Explore: Phân tích dữ liệu để tìm mẫu và mối quan hệ.

  - Modify: Chuyển đổi dữ liệu để phù hợp với mô hình.

  - Model: Áp dụng thuật toán để xây dựng mô hình.

  - Assess: Đánh giá hiệu quả mô hình.

**Khác biệt chính:** CRISP-DM tập trung vào quy trình nghiệp vụ toàn diện, còn SEMMA thiên về kỹ thuật phân tích dữ liệu.

## **Câu hỏi 2**

**Cây quyết định hoạt động như thế nào? Hãy giải thích các thành phần chính (nút gốc, nút lá, nhánh)
và cách cây đưa ra dự đoán.**

### **Trả lời**

Cây quyết định là mô hình phân loại hoặc hồi quy, hoạt động bằng cách chia dữ liệu thành các nhánh dựa trên thuộc tính.

**Các thành phần chính:**
  - Nút gốc (Root Node): Nút đầu tiên, chứa toàn bộ dữ liệu.

  - Nhánh (Branch): Đại diện cho điều kiện phân chia dữ liệu.

  - Nút lá (Leaf Node): Nút cuối cùng, chứa kết quả phân loại hoặc dự đoán.

**Cách dự đoán:**
Dữ liệu mới được đưa vào cây, đi qua các nhánh theo điều kiện cho đến khi đến nút lá, nơi đưa ra kết quả dự đoán.

## **Câu hỏi 3**

**Các tiêu chí phân tách (splitting criteria) như Gini Index, Entropy, hay Information Gain được sử dụng
trong cây quyết định là gì? Chúng khác nhau ra sao?**

### **Trả lời**

**Gini Index**
  - Đo độ tinh khiết của nút.

  - Công thức: $Gini = 1 - \sum_{i=1}^{n} p_i^2$

  - Giá trị càng thấp → nút càng thuần.

**Entropy**
  - Đo độ hỗn loạn của dữ liệu.

  - Công thức: $Entropy = - \sum_{i=1}^{n} p_i \log_2 p_i$

  - Giá trị càng cao → dữ liệu càng hỗn loạn.

**Information Gain**
  - Đo mức giảm entropy sau khi phân chia.

  - Công thức: $IG = Entropy_{parent} - \sum_{j=1}^{k} \frac{n_j}{n} \cdot Entropy_j$

  - Dùng để chọn thuộc tính phân chia tốt nhất.

## **Câu hỏi 4**

**Rừng cây (Random Forest) là gì? Nó khác gì so với một cây quyết định đơn lẻ? Tại sao Random Forest thường có hiệu suất tốt hơn cây quyết định trong các bài toán phân loại?**

### **Trả lời**

**1. Random Forest là gì?**
Random Forest là một thuật toán học máy thuộc nhóm ensemble learning, kết hợp nhiều cây quyết định (decision trees) để đưa ra dự đoán cuối cùng bằng cách bỏ phiếu (với phân loại) hoặc trung bình (với hồi quy).

**Đặc điểm chính:**
  - Mỗi cây được huấn luyện trên một tập con ngẫu nhiên của dữ liệu (bootstrap sampling).

  - Tại mỗi nút phân chia, chỉ một tập con ngẫu nhiên các đặc trưng được xét đến.

  - Dự đoán cuối cùng là tổng hợp từ nhiều cây → giảm sai số.

**2. Khác biệt giữa Random Forest và cây quyết định đơn lẻ**
| **Tiêu chí**              | **Cây quyết định đơn lẻ**         | **Random Forest**                          |
|--------------------------|-----------------------------------|--------------------------------------------|
| Cấu trúc                 | Một cây duy nhất                  | Tập hợp nhiều cây quyết định               |
| Dễ bị quá khớp           | Cao                               | Thấp (nhờ tổng hợp nhiều cây)              |
| Độ chính xác             | Phụ thuộc vào cây duy nhất        | Cao hơn nhờ trung bình hóa kết quả         |
| Tính diễn giải           | Dễ hiểu và trực quan              | Khó hơn (nhiều cây, không trực quan)       |
| Tính ổn định             | Nhạy cảm với dữ liệu              | Ổn định hơn khi dữ liệu thay đổi nhẹ       |
| Khả năng chống nhiễu     | Thấp                              | Cao (nhờ đa dạng hóa dữ liệu và đặc trưng) |

**3. Vì sao Random Forest thường hiệu quả hơn?**
  - Giảm overfitting: Một cây đơn lẻ dễ học quá kỹ dữ liệu huấn luyện, còn Random Forest trung bình hóa kết quả từ nhiều cây → giảm sai số.

  - Tăng độ chính xác: Nhờ tổng hợp nhiều mô hình, Random Forest thường cho kết quả tốt hơn trong thực tế.

  - Ổn định: Ít bị ảnh hưởng bởi nhiễu hoặc thay đổi nhỏ trong dữ liệu.

  - Làm việc tốt với dữ liệu thiếu hoặc nhiễu: Vì mỗi cây chỉ dùng một phần dữ liệu và đặc trưng.

## **Câu hỏi 5**

**Những ưu điểm và hạn chế của cây quyết định và Random Forest là gì? Trong trường hợp nào thì cây
quyết định có thể hoạt động kém hiệu quả?**

### **Trả lời**

**Cây quyết định (Decision Tree)**
- **Ưu điểm:**
  - Dễ hiểu, dễ diễn giải (có thể vẽ sơ đồ cây).

  - Không cần chuẩn hóa dữ liệu.

  - Có thể xử lý cả dữ liệu số và phân loại.

  - Tính toán nhanh, phù hợp với bài toán nhỏ.

- **Hạn chế:**
  - Dễ bị quá khớp (overfitting) nếu cây quá sâu.

  - Nhạy cảm với dữ liệu nhiễu hoặc thay đổi nhỏ.

  - Hiệu suất thấp với dữ liệu phức tạp hoặc nhiều chiều.


**Random Forest**
- **Ưu điểm:**
  - Giảm overfitting nhờ trung bình hóa nhiều cây.

  - Hiệu suất cao trong nhiều bài toán phân loại và hồi quy.

  - Ổn định hơn khi dữ liệu có nhiễu.

  - Có thể đánh giá tầm quan trọng của đặc trưng (feature importance).

- **Hạn chế:**
  - Khó diễn giải (do có nhiều cây).

  - Tốn tài nguyên tính toán hơn cây đơn lẻ.

  - Không phù hợp nếu cần mô hình đơn giản, dễ giải thích.
  

**Khi nào cây quyết định hoạt động kém hiệu quả?**
  - Khi dữ liệu có nhiều nhiễu hoặc biến không liên quan.

  - Khi dữ liệu có nhiều chiều (high-dimensional) → dễ bị quá khớp.

  - Khi yêu cầu mô hình có tính tổng quát cao hoặc hiệu suất cao → Random Forest sẽ tốt hơn.

  - Khi dữ liệu có sự phân bố không đều hoặc thiếu mẫu ở một số lớp.

## **Câu hỏi 6**

**Viết đoạn code mẫu bằng Python (sử dụng Scikit-learn) để xây dựng một mô hình cây quyết định
không? Hãy mô tả các bước thực hiện**

### **Trả lời**

**Các bước thực hiện:**
  1. Import thư viện cần thiết

  2. Tải dữ liệu mẫu (ở đây dùng Iris dataset)

  3. Tiền xử lý dữ liệu: chia tập huấn luyện và kiểm tra

  4. Huấn luyện mô hình cây quyết định

  5. Dự đoán và đánh giá

  6. Vẽ cây quyết định

```python
# Bước 1: Import thư viện
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Bước 2: Tải dữ liệu mẫu
iris = load_iris()
X = iris.data
y = iris.target

# Bước 3: Chia dữ liệu thành tập huấn luyện và kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Bước 4: Khởi tạo và huấn luyện mô hình cây quyết định
model = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=42)
model.fit(X_train, y_train)

# Bước 5: Dự đoán và đánh giá
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Độ chính xác của mô hình: {accuracy:.2f}")

# Bước 6: Vẽ cây quyết định
plt.figure(figsize=(12, 8))
plot_tree(model, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.title("Cây quyết định - Iris Dataset")
plt.show()
```

## **Câu hỏi 7**

**Làm thế nào để triển khai một mô hình Random Forest trong Python? Bạn thường thiết lập các tham
số nào (ví dụ: n_estimators, max_depth)?**

### **Trả lời**

**1. Cài đặt thư viện cần thiết**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

**2. Tải và chia dữ liệu**

```python
# Dữ liệu mẫu: Iris
X, y = load_iris(return_X_y=True)

# Chia dữ liệu thành tập huấn luyện và kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

**3. Khởi tạo và huấn luyện mô hình Random Forest**

```python
model = RandomForestClassifier(
    n_estimators=100,     # Số lượng cây trong rừng
    max_depth=5,          # Độ sâu tối đa của mỗi cây
    random_state=42       # Đảm bảo kết quả có thể tái lập
)
model.fit(X_train, y_train)
```

**4. Dự đoán và đánh giá**

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Độ chính xác: {accuracy:.2f}")
```

**5. Các tham số thường được thiết lập**

| **Tham số**           | **Ý nghĩa**                                                                 |
|-----------------------|------------------------------------------------------------------------------|
| n_estimators        | Số lượng cây trong rừng (mặc định: 100). Tăng số lượng giúp ổn định hơn.     |
| max_depth           | Độ sâu tối đa của mỗi cây. Giới hạn để tránh overfitting.                    |
| min_samples_split  | Số mẫu tối thiểu để chia một nút (mặc định: 2).                              |
| min_samples_leaf    | Số mẫu tối thiểu ở một nút lá.                                               |
| max_features        | Số lượng đặc trưng được xét tại mỗi lần chia (mặc định: 'sqrt').             |
| bootstrap           | Có lấy mẫu lại dữ liệu hay không (True/False).                               |
| random_state        | Đảm bảo kết quả có thể tái lập.                                              |


## **Câu hỏi 8**

**Làm thế nào để đánh giá tầm quan trọng của các đặc trưng (feature importance) trong Random
Forest bằng Python?**

### **Trả lời**

**Các bước thực hiện:**

   **1. Huấn luyện mô hình Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris

# Tải dữ liệu mẫu
X, y = load_iris(return_X_y=True)
feature_names = load_iris().feature_names

# Huấn luyện mô hình
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)
```

**2. Lấy giá trị độ quan trọng của đặc trưng**

```python
import pandas as pd

# Lấy độ quan trọng
importances = model.feature_importances_

# Hiển thị dưới dạng bảng
df_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

print(df_importance)
```

**3. Vẽ biểu đồ trực quan**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.barh(df_importance['Feature'], df_importance['Importance'], color='skyblue')
plt.xlabel("Feature Importance")
plt.title("Độ quan trọng của các đặc trưng")
plt.gca().invert_yaxis()
plt.show()
```

## **Câu hỏi 9**

**Điều chỉnh siêu tham số (hyperparameter tuning) cho cây quyết định hoặc Random Forest chưa? Hãy
mô tả cách bạn sử dụng GridSearchCV hoặc RandomizedSearchCV**

### **Trả lời**

Điều chỉnh siêu tham số (hyperparameter tuning) là quá trình tìm ra tổ hợp tham số tối ưu để cải thiện hiệu suất mô hình. Với cây quyết định hoặc Random Forest, bạn có thể sử dụng GridSearchCV hoặc RandomizedSearchCV trong thư viện Scikit-learn để thực hiện việc này.

**GridSearchCV vs RandomizedSearchCV**
| **Phương pháp**         | **Đặc điểm**                                                                 |
|-------------------------|------------------------------------------------------------------------------|
| GridSearchCV          | Thử tất cả tổ hợp tham số trong lưới → chính xác nhưng tốn thời gian         |
| RandomizedSearchCV    | Chọn ngẫu nhiên một số tổ hợp → nhanh hơn, phù hợp với tập tham số lớn       |


**Các bước thực hiện điều chỉnh siêu tham số**

**1. Import thư viện cần thiết**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

**2. Định nghĩa mô hình và tập tham số**

```python
model = RandomForestClassifier(random_state=42)

param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2']
}
```

**3. Khởi tạo GridSearchCV hoặc RandomizedSearchCV**

```python
# GridSearchCV
grid_search = GridSearchCV(model, param_grid, cv=5, scoring='accuracy')

# Hoặc dùng RandomizedSearchCV
# from sklearn.model_selection import RandomizedSearchCV
# random_search = RandomizedSearchCV(model, param_grid, n_iter=10, cv=5, scoring='accuracy', random_state=42)
```

**4. Huấn luyện và tìm tham số tốt nhất**

```python
grid_search.fit(X_train, y_train)
print("Best parameters:", grid_search.best_params_)
print("Best score:", grid_search.best_score_)
```
