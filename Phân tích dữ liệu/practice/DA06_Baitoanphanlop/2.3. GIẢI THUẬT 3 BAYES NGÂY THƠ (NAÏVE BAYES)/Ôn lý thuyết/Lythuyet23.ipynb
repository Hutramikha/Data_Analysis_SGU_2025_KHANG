## **Câu hỏi 1**

**Giải thuật Naive Bayes hoạt động như thế nào? Hãy giải thích định lý Bayes và giả định "ngây thơ"
trong thuật toán này?**

### **Trả lời**

**Naive Bayes** là một thuật toán phân loại dựa trên định lý Bayes, với giả định “ngây thơ” rằng các đặc trưng (features) là độc lập có điều kiện với nhau khi biết nhãn lớp. Dù giả định này hiếm khi đúng trong thực tế, Naive Bayes vẫn hoạt động rất hiệu quả trong nhiều bài toán như phân loại văn bản, lọc spam, và phân tích cảm xúc.

**Định lý Bayes là gì?**

Định lý Bayes mô tả xác suất xảy ra của một sự kiện dựa trên kiến thức có trước về các điều kiện liên quan đến sự kiện đó:
$$
P(C|X) = \frac{P(X|C) \cdot P(C)}{P(X)}
$$

Trong đó:
- \( P(C|X) \): xác suất lớp \( C \) xảy ra khi biết đặc trưng \( X \)
- \( P(X|C) \): xác suất quan sát thấy đặc trưng \( X \) khi biết lớp \( C \)
- \( P(C) \): xác suất tiên nghiệm của lớp \( C \)
- \( P(X) \): xác suất quan sát thấy đặc trưng \( X \)


**Giả định "ngây thơ" (Naive Assumption)**

Thuật toán giả định rằng các đặc trưng là độc lập với nhau khi biết nhãn lớp:

$$
P(X|C) = P(x_1|C) \cdot P(x_2|C) \cdot \dots \cdot P(x_n|C)
$$

=> Giả định rằng các đặc trưng \( x_1, x_2, \dots, x_n \) là **độc lập có điều kiện** khi biết lớp \( C \)


## **Câu hỏi 2**

**Các loại mô hình Naive Bayes (Gaussian, Multinomial, Bernoulli) khác nhau ra sao? Khi nào nên sử dụng
từng loại?**

### **Trả lời**

**Naive Bayes** có ba loại chính, mỗi loại phù hợp với kiểu dữ liệu khác nhau:

- Gaussian Naive Bayes
  - Giả định đặc trưng có phân phối chuẩn (Gaussian).

  - Dùng cho dữ liệu liên tục (float).

  - Ví dụ: chiều cao, nhiệt độ, điểm số.

- Multinomial Naive Bayes
  - Giả định đặc trưng tuân theo phân phối đa thức (Multinomial).

  - Dùng cho dữ liệu rời rạc, dạng đếm (số nguyên ≥ 0).

  - Ví dụ: số lần từ xuất hiện trong văn bản (Bag-of-Words, TF-IDF).

- Bernoulli Naive Bayes
  - Giả định đặc trưng là nhị phân (0 hoặc 1).

  - Dùng cho dữ liệu nhị phân, có/không, đúng/sai.

  - Ví dụ: từ có xuất hiện hay không trong văn bản.

- Khi nào dùng từng loại:

  - Gaussian → dữ liệu liên tục.

  - Multinomial → dữ liệu đếm, phân loại văn bản.

  - Bernoulli → dữ liệu nhị phân, lọc spam hoặc phân loại có/không.

## **Câu hỏi 3**

**Tại sao Naive Bayes được gọi là "ngây thơ"? Giả định về tính độc lập của các đặc trưng ảnh hưởng như
thế nào đến hiệu suất của mô hình?**

### **Trả lời**

**Naive Bayes** được gọi là “ngây thơ” vì nó giả định rằng các đặc trưng (features) là độc lập với nhau có điều kiện khi biết nhãn lớp — tức là mô hình cho rằng sự xuất hiện của một đặc trưng không ảnh hưởng đến đặc trưng khác, điều này hiếm khi đúng trong thực tế.

**Giả định “ngây thơ” là gì?**

Với một tập đặc trưng: $$ X = \{x_1, x_2, \dots, x_n\} $$
Naive Bayes giả định: $$ P(X \mid C) = \prod_{i=1}^{n} P(x_i \mid C) $$

=> Giả định rằng các đặc trưng \( x_1, x_2, \dots, x_n \) là độc lập có điều kiện khi biết lớp \( C \)

Vậy, điều này giúp đơn giản hóa việc tính toán xác suất, giảm chi phí tính toán và dễ triển khai.

**Ảnh hưởng đến hiệu suất mô hình**
- **Tích cực:**

  - Mô hình nhanh, đơn giản, dễ huấn luyện ngay cả với dữ liệu lớn.

  - Hoạt động tốt trong các bài toán như phân loại văn bản, nơi đặc trưng (từ) thường gần như độc lập.

- **Tiêu cực:**

  - Nếu các đặc trưng phụ thuộc mạnh lẫn nhau, giả định này có thể dẫn đến dự đoán sai lệch.

  - Mô hình có thể giảm độ chính xác nếu không xử lý tốt mối quan hệ giữa các đặc trưng.

**Tóm lại:** Naive Bayes “ngây thơ” vì giả định độc lập giữa các đặc trưng. Dù không thực tế, giả định này giúp mô hình đơn giản và hiệu quả, nhưng có thể ảnh hưởng đến độ chính xác nếu đặc trưng phụ thuộc lẫn nhau.

## **Câu hỏi 4**

**Ưu điểm và hạn chế của Naive Bayes so với các thuật toán phân loại khác như SVM hoặc Random Forest
là gì?**

### **Trả lời**

- **Ưu điểm của Naive Bayes**
  - Huấn luyện nhanh, đơn giản, hiệu quả với dữ liệu lớn.

  - Tốt với dữ liệu rời rạc như văn bản (Bag-of-Words, TF-IDF).

  - Ít bị overfitting, hoạt động tốt với dữ liệu nhỏ.

- **Hạn chế của Naive Bayes**
  - Giả định đặc trưng độc lập không thực tế → giảm độ chính xác.

  - Không xử lý tốt mối quan hệ phi tuyến giữa đặc trưng.

  - Hiệu suất kém với dữ liệu có đặc trưng phụ thuộc mạnh.

- **So sánh Naive Bayes với SVM và Random Forest**

| Tiêu chí              | Naive Bayes                     | SVM                                  | Random Forest                        |
|----------------------|----------------------------------|--------------------------------------|--------------------------------------|
| Tốc độ huấn luyện     | Rất nhanh                       | Chậm hơn                             | Trung bình đến chậm                  |
| Độ chính xác          | Tốt với dữ liệu đơn giản        | Cao với dữ liệu phức tạp             | Cao, tổng quát tốt                   |
| Xử lý đặc trưng       | Rời rạc, độc lập                | Liên tục, tuyến tính hoặc kernel     | Rời rạc và liên tục                  |
| Nhạy cảm với nhiễu    | Ít nhạy cảm                     | Có thể bị ảnh hưởng nếu không tối ưu | Khá ổn định                          |
| Giải thích mô hình    | Dễ hiểu                         | Khó giải thích                       | Khó giải thích                       |
| Khả năng mở rộng      | Cao                             | Cần điều chỉnh kernel và tham số     | Tốt, nhưng tốn tài nguyên            |


- **Tóm lại:**

  - Naive Bayes phù hợp với dữ liệu rời rạc, văn bản, bài toán đơn giản.

  - SVM mạnh với dữ liệu phức tạp, cần tối ưu kernel.

  - Random Forest linh hoạt, chính xác cao nhưng tốn tài nguyên.

## **Câu hỏi 5**

**Viết đoạn code mẫu bằng Python (sử dụng Scikit-learn) để xây dựng một mô hình Naive Bayes (ví dụ:
Gaussian Naive Bayes) không? Hãy mô tả các bước thực hiện**

### **Trả lời**

```python
# Bước 1: Import thư viện
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# Bước 2: Tải dữ liệu
X, y = load_iris(return_X_y=True)

# Bước 3: Chia dữ liệu thành tập huấn luyện và kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Bước 4: Chuẩn hóa dữ liệu (tùy chọn nhưng nên làm)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Bước 5: Khởi tạo và huấn luyện mô hình Gaussian Naive Bayes
model = GaussianNB()
model.fit(X_train_scaled, y_train)

# Bước 6: Dự đoán và đánh giá
y_pred = model.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

## **Câu hỏi 6**

**Làm thế nào để xử lý dữ liệu phân loại (categorical data) trước khi áp dụng Multinomial Naive Bayes
trong Python?**

### **Trả lời**

**Các bước xử lý dữ liệu phân loại**

| Bước                     | Mô tả ngắn gọn                                      |
|--------------------------|-----------------------------------------------------|
| 1. Mã hóa dữ liệu         | Dùng OneHotEncoder hoặc CountEncoder để chuyển đổi |
| 2. Chuyển sang số nguyên | Đảm bảo dữ liệu là số nguyên ≥ 0                   |
| 3. Tách dữ liệu           | Dùng train_test_split để chia tập huấn luyện/test  |
| 4. Huấn luyện mô hình     | Dùng MultinomialNB từ sklearn.naive_bayes          |


**Ví dụ mã hóa bằng OneHotEncoder**

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split

# Dữ liệu mẫu
X = [['red'], ['green'], ['blue'], ['green'], ['red']]
y = [0, 1, 2, 1, 0]

# Bước 1: Mã hóa dữ liệu phân loại
encoder = OneHotEncoder()
X_encoded = encoder.fit_transform(X).toarray()

# Bước 2: Tách dữ liệu
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.3, random_state=42)

# Bước 3: Huấn luyện Multinomial Naive Bayes
model = MultinomialNB()
model.fit(X_train, y_train)
```

**Lưu ý:**

  - MultinomialNB yêu cầu dữ liệu là số nguyên không âm.

  - Nếu dữ liệu là văn bản, dùng CountVectorizer hoặc TfidfVectorizer.

## **Câu hỏi 7**

**Naive Bayes thường được sử dụng trong phân loại văn bản (text classification). Bạn có thể giải thích
cách triển khai Naive Bayes cho bài toán này không?**

### **Trả lời**

**Các bước triển khai Naive Bayes cho phân loại văn bản**
| Bước                     | Mô tả ngắn gọn                                                       |
|--------------------------|----------------------------------------------------------------------|
| 1. Thu thập dữ liệu       | Tập văn bản có nhãn (ví dụ: spam/ham, tích cực/tiêu cực)            |
| 2. Tiền xử lý văn bản     | Làm sạch, chuyển về chữ thường, loại bỏ stopwords, stemming         |
| 3. Vector hóa văn bản     | Dùng CountVectorizer hoặc TfidfVectorizer để chuyển văn bản thành số |
| 4. Chia tập dữ liệu       | Dùng train_test_split để chia train/test                             |
| 5. Huấn luyện mô hình     | Dùng MultinomialNB từ sklearn.naive_bayes                            |
| 6. Dự đoán và đánh giá    | Dùng accuracy_score, classification_report để kiểm tra hiệu suất     |


**Ví dụ code ngắn gọn**

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Dữ liệu mẫu
texts = ["I love this movie", "This film is terrible", "Amazing plot", "Worst acting ever"]
labels = [1, 0, 1, 0]  # 1: tích cực, 0: tiêu cực

# Bước 1–3: Vector hóa văn bản
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# Bước 4: Chia dữ liệu
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.3, random_state=42)

# Bước 5: Huấn luyện mô hình
model = MultinomialNB()
model.fit(X_train, y_train)

# Bước 6: Dự đoán và đánh giá
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
